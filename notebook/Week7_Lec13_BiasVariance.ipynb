## 1. Introduction

In machine learning, one of the key challenges is ensuring that the model generalizes well to unseen data. The concepts of **over-fitting**, **under-fitting**, and the **bias-variance tradeoff** are central to understanding model performance.

### Underfitting

Under-fitting happens when a model is too simple to capture the underlying patterns in the data. As a result, it performs poorly both on the training data and new data. This typically occurs when the model has **high bias** and **low variance**.

- **Bias**: Bias refers to the error introduced by the model's assumptions. High bias means the model is making strong assumptions about the data, leading to **under-fitting**.

Consider fitting a linear model to data generated by a complex non-linear function. A linear model is too simple and thus underfits the data.

### Overfitting

Over-fitting occurs when a model learns not only the underlying patterns but also the noise in the training data. This results in **high variance** and **low bias**, where the model performs well on the training data but poorly on new data.

- **Variance**: Variance refers to how much the model's predictions would change if we trained it on a different training set. A model with high variance is too sensitive to the training data, causing it to overfit.

A model with too many parameters (like a high-degree polynomial) can fit every detail of the training data, including noise, leading to poor generalization on new data.

## 2. Mathematical Representation of Bias and Variance

In machine learning, the **expected prediction error** of a model can be decomposed into three components: **bias**, **variance**, and **irreducible error**. This is known as the **bias-variance decomposition**, and it is a critical concept for understanding model performance and generalization.

### 2.1 Bias and Variance Mathematical Definition

The **bias** measures how far off the model’s predictions are from the true values. It is the error introduced by the model's assumptions about the underlying data. A model with high bias pays little attention to the training data and oversimplifies the problem, which can lead to **underfitting**.

The bias is defined as the difference between the expected prediction of the model and the true value:
$$
\text{Bias}(x_0) = \mathbb{E}[\hat{f}(x_0)] - f(x_0)
$$
- $\hat{f}(x_0)$ is the predicted value by the model at point $x_0$.
- $f(x_0)$ is the true function value at point $x_0$.
- $\mathbb{E}[\hat{f}(x_0)]$ is the expected value of the model's predictions.

The **squared bias** is:
$$
\text{Bias}^2(x_0) = (\mathbb{E}[\hat{f}(x_0)] - f(x_0))^2
$$

The **variance** measures the variability of the model’s predictions for a given data point when trained on different datasets. A model with high variance pays too much attention to the training data and may overfit the model, leading to poor generalization to unseen data.

The variance is defined as:
$$
\text{Var}(x_0) = \mathbb{E}[(\hat{f}(x_0) - \mathbb{E}[\hat{f}(x_0)])^2]
$$
- $\hat{f}(x_0)$ is the model's prediction at point $x_0$.
- $\mathbb{E}[\hat{f}(x_0)]$ is the expected prediction across different datasets.

  
### 2.2 Expected Prediction Error (total error)

Let’s assume we have a dataset $\mathcal{D} = \{(x_i, y_i)\}_{i=1}^n $ where:
- $x_i$ are the input features
- $y_i$ are the corresponding target values

We want to fit a model $\hat{f}(x)$ to predict the true function $f(x)$, but the observed values $y$ have some inherent noise:

$$
y = f(x) + \epsilon
$$
where $\epsilon$ is a random noise term with zero mean, i.e., $\mathbb{E}[\epsilon] = 0$ , and variance $\sigma^2_{\epsilon}$, i.e., $ \text{Var}(\epsilon) = \sigma^2_{\epsilon}$.

The goal is to minimize the **expected prediction error** at a new point $x_0$, which is given by the **mean squared error (MSE)**:

$$
\text{Total Error (MSE)}=\mathbb{E}[(\hat{f}(x_0) - y_0)^2]
$$

where $y_0 = f(x_0) + \epsilon$. The **expected prediction error** can be decomposed into three terms: **bias**, **variance**, and **irreducible error**.


#### Step 1: Expand the MSE

The MSE can be rewritten as:

$$
\mathbb{E}[(\hat{f}(x_0) - y_0)^2] = \mathbb{E}[(\hat{f}(x_0) - f(x_0) - \epsilon)^2]
$$

Since $y_0 = f(x_0) + \epsilon$, we can substitute and expand the squared term:

$$
= \mathbb{E}[(\hat{f}(x_0) - f(x_0))^2 + 2(\hat{f}(x_0) - f(x_0))(-\epsilon) + \epsilon^2]
$$

Since $\mathbb{E}[\epsilon] = 0$, the second term vanishes, leaving us with:

$$
= \mathbb{E}[(\hat{f}(x_0) - f(x_0))^2] + \mathbb{E}[\epsilon^2]
$$

Because $\mathbb{E}[\epsilon^2] = \sigma^2_{\epsilon}$, we now have:

$$
\mathbb{E}[(\hat{f}(x_0) - y_0)^2] = \underbrace{\mathbb{E}[(\hat{f}(x_0) - f(x_0))^2]}_{\text{Model Error}} + \underbrace{\sigma^2_{\epsilon}}_{\text{Irreducible Error (noise)}}
$$

This shows that the total error can be decomposed into two parts:
1. The error due to the model $\hat{f}(x_0)$ not perfectly fitting $f(x_0)$
2. The irreducible error $\sigma^2_{\epsilon}$ due to inherent noise in the data

#### Step 2: Decompose the Model Error

Now, decompose the first term $\mathbb{E}[(\hat{f}(x_0) - f(x_0))^2]$ into **bias** and **variance** components. First, rewrite the model’s prediction as:

$$
\mathbb{E}[(\hat{f}(x_0) - f(x_0))^2] = \mathbb{E}[(\hat{f}(x_0) - \mathbb{E}[\hat{f}(x_0)] + \mathbb{E}[\hat{f}(x_0)] - f(x_0))^2]
$$

Expanding the square:

$$
\text{Model Error}= \mathbb{E}[(\hat{f}(x_0) - \mathbb{E}[\hat{f}(x_0)])^2] + (\mathbb{E}[\hat{f}(x_0)] - f(x_0))^2 + 2\mathbb{E}[(\hat{f}(x_0) - \mathbb{E}[\hat{f}(x_0)])(\mathbb{E}[\hat{f}(x_0)] - f(x_0))]
$$

The cross-term vanishes because $\mathbb{E}[\hat{f}(x_0)] - f(x_0)$ is constant with respect to the expectation over the dataset, so we are left with:

$$
\text{Model Error}= \underbrace{\mathbb{E}[(\hat{f}(x_0) - \mathbb{E}[\hat{f}(x_0)])^2]}_{\text{Variance}} + \underbrace{(\mathbb{E}[\hat{f}(x_0)] - f(x_0))^2}_{\text{Bias}^2}
$$

Thus, the total expected error becomes:

\begin{align}
\mathbb{E}[(\hat{f}(x_0) - y_0)^2] (MSE) &= \text{Model Error} + \text{Irreducible Error} \\
 &= \underbrace{(\mathbb{E}[\hat{f}(x_0)] - f(x_0))^2}_{\text{Bias}^2} + \underbrace{\mathbb{E}[(\hat{f}(x_0) - \mathbb{E}[\hat{f}(x_0)])^2]}_{\text{Variance}} + \underbrace{\sigma^2_{\epsilon}}_{\text{Irreducible Error}}
\end{align}

- **Bias** ($\text{Bias}^2$) measures the error introduced by approximating the true function $f(x)$ with a simplified model. High bias typically occurs when the model is too simple to capture the underlying structure of the data (e.g., underfitting).
  
- **Variance** measures the sensitivity of the model to different training sets. A high variance model will change significantly with small variations in the training data, leading to overfitting.

- **Irreducible error** $\sigma^2_{\epsilon}$ is due to the inherent noise in the data that cannot be eliminated, no matter how good the model is.
- The goal of model training is to minimize both the bias and variance.

## 3. Trade-off Between Bias and Variance

The goal is to find a balance between bias and variance, which results in the lowest possible total error:
$$
\text{Total Error} = \text{Bias}^2 + \text{Variance} + \text{Irreducible Error}
$$

There is a fundamental trade-off between bias and variance:
- **High Bias, Low Variance**: The model is simple and makes strong assumptions (e.g., linear regression). It has poor performance on both training and test data (under-fitting).
- **Low Bias, High Variance**: The model is too complex and adapts too much to the training data (e.g., deep neural networks, high-order degree polynomial, high-order kernels etc.). It performs well on the training data but poorly on the test data (over-fitting).

### Summary
- The **bias-variance decomposition** provides valuable insight into the sources of error in machine learning models. By understanding the trade-off between bias and variance, we can better select models and fine-tune hyperparameters to improve model performance and generalization.
- To build a robust model, it's crucial to find the right balance between bias and variance. Too simple a model (high bias) will underfit the data, while too complex a model (high variance) will overfit. Cross-validation and regularization techniques can help control these issues and lead to better generalization.
